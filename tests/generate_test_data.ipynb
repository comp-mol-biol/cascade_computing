```python
#pick chains
sys_domains=pd.read_parquet('../examples/data/replica_6/postprocessing/test_project_MUT16_65_46af39bf2445d01c22c25aadf9d305c4_full_pi_ref_debug_10_chains_2001.xtc')
ALL_MOLS_SEL = [f"resid {row['min']}-{row['max']}" for _, row in sys_domains.iterrows()]
combinations_in = list(itertools.combinations(enumerate(ALL_MOLS_SEL), 2))
list_comb=combinations_in[10:20]
```

```python
#make chain selection
l=[]
for item in list_comb:
    #print(item[0][1], item[1][1])
    l.append(item[0][1])
    l.append(item[1][1])

selection_string = " or ".join(set(l))
sel_10_chains=u.select_atoms(selection_string)
sel_10_chains
```
